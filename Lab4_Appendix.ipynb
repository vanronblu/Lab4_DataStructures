```python
# Lab 4 Appendix 
# Student-specific inputs
LAST_NAME = "GERMAN"
SEED_NUM = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

seed_digits = [int(ch) for ch in SEED_NUM if ch.isdigit()]          # [3,6,4,3,3,6]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]     # U,N,I,Q,...

sources = [LAST_NAME + "_T" + str(n) for n in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

def make_value(i):
    digit = seed_digits[i % len(seed_digits)]
    letter = artist_letters[i]
    return digit * (i + 1) + (ord(letter) - 64)

# (source index, category index) for each of the 10 records
plan = [(0, 0), (1, 1), (2, 2), (3, 3), (4, 4),   # 5 base records
        (0, 0),                                    # repeated source/category
        (0, 1), (2, 1),                            # extra VOLTAGE measurements
        (3, 2), (4, 2)]                            # extra CURRENT measurements

#  Generated Test Data (list of tuples) 
records = []
for i, (s, c) in enumerate(plan):
    records.append((sources[s], categories[c], make_value(i)))

print("=== 1. GENERATED TEST DATA ===")
print("No.  Source        Category      Value")
for n, (s, c, v) in enumerate(records, 1):
    print(f"{n:<4} {s:<13} {c:<13} {v}")

# Organized Test Data (dictionary: source/category/values) 
organized = {}
for s, c, v in records:
    if s not in organized:
        organized[s] = {}
    if c not in organized[s]:
        organized[s][c] = []
    organized[s][c].append(v)

print("\n=== 2. ORGANIZED TEST DATA (by source and category) ===")
for s in organized:
    print(s)
    for c in organized[s]:
        print("   ", c, "->", organized[s][c])

# Repeated / Distinct Data 
seen = set()
repeated = []
for s, c, v in records:
    if (s, c) in seen and (s, c) not in repeated:
        repeated.append((s, c))
    seen.add((s, c))

distinct_sources = set(s for s, c, v in records)
distinct_categories = set(c for s, c, v in records)
distinct_pairs = seen

print("\n=== 3. REPEATED / DISTINCT DATA ===")
print("Repeated source/category:", repeated)
print("Distinct sources   :", len(distinct_sources), sorted(distinct_sources))
print("Distinct categories:", len(distinct_categories), sorted(distinct_categories))
print("Distinct source/category pairs:", len(distinct_pairs), "of", len(records), "records")

# Numerical and Categorical Summaries 
category_values = {}
for s, c, v in records:
    category_values.setdefault(c, []).append(v)

print("\n=== 4. NUMERICAL AND CATEGORICAL SUMMARIES ===")
all_values = [v for s, c, v in records]
print("Overall -> count:", len(all_values), "| total:", sum(all_values),
      "| min:", min(all_values), "| max:", max(all_values),
      "| average:", round(sum(all_values) / len(all_values), 2))
print("\nPer category:")
for c in categories:
    vals = category_values[c]
    print(f"  {c:<12} count={len(vals)} total={sum(vals)} "
          f"min={min(vals)} max={max(vals)} avg={round(sum(vals)/len(vals), 2)}")

source_totals = {}
for s, c, v in records:
    source_totals[s] = source_totals.get(s, 0) + v
print("\nPer source (total value):")
for s in sources:
    print(" ", s, "=", source_totals[s])

# Processed Results
average_all = sum(all_values) / len(all_values)
processed = []
for s, c, v in records:
    level = "ABOVE AVERAGE" if v > average_all else "BELOW/EQUAL AVERAGE"
    processed.append((s, c, v, level))

print("\n=== 5. PROCESSED RESULTS (compared with overall average",
      round(average_all, 2), ") ===")
for s, c, v, level in processed:
    print(f"  {s:<13} {c:<12} {v:<4} {level}")

# Ordered Results 
by_value = sorted(records, key=lambda r: r[2], reverse=True)
by_source_category = sorted(records, key=lambda r: (r[0], r[1]))

print("\n=== 6. ORDERED RESULTS ===")
print("Highest to lowest value:")
for s, c, v in by_value:
    print(f"  {v:<4} {s:<13} {c}")
print("\nBy source then category:")
for s, c, v in by_source_category:
    print(f"  {s:<13} {c:<12} {v}")

# Final Summary / Final Organized and Summarized Test Dataset 
print("\n" + "=" * 62)
print("FINAL ORGANIZED AND SUMMARIZED TEST DATASET")
print("=" * 62)
print("Student : ", LAST_NAME, "|", SEED_NUM, "|", FAVORITE_ARTIST)
print("Total records       :", len(records))
print("Distinct sources    :", len(distinct_sources))
print("Distinct categories :", len(distinct_categories))
print("Repeated record     :", repeated)
print("Overall total value :", sum(all_values))
print("Overall average     :", round(average_all, 2))
print("Highest measurement :", by_value[0])
print("Lowest measurement  :", by_value[-1])
print("-" * 62)
print(f"{'SOURCE':<13} {'CATEGORY':<12} {'VALUES':<14} {'TOTAL'}")
for s in sorted(organized):
    for c in organized[s]:
        vals = organized[s][c]
        print(f"{s:<13} {c:<12} {str(vals):<14} {sum(vals)}")
print("=" * 62)
```

```python
# Lab 4 Appendix - Additional Exercise 2: Student Performance Data Analyzer
# Student-specific inputs
LAST_NAME = "GERMAN"
SEED_NUM = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

seed_digits = [int(ch) for ch in SEED_NUM if ch.isdigit()]          # [3,6,4,3,3,6]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]     # U,N,I,Q,...

students = [LAST_NAME + "_S" + str(n) for n in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def make_score(i):
    digit = seed_digits[i % len(seed_digits)]
    letter = artist_letters[i % len(artist_letters)]
    return 40 + (digit * (i + 1) + (ord(letter) - 64)) % 60

def get_category(score):
    if score >= 90:
        return "EXCELLENT"
    elif score >= 85:
        return "VERY GOOD"
    elif score >= 75:
        return "GOOD"
    elif score >= 60:
        return "FAIR"
    else:
        return "POOR"

# (student index, subject index, has_score)
plan = [(0, 0, True), (1, 1, True), (2, 2, True), (3, 3, True), (4, 4, True),  # base records
        (0, 0, True),                                                          # repeated student-subject
        (0, 1, True), (1, 2, True), (2, 3, True), (3, 4, True), (4, 0, True),
        (1, 3, False)]                                                         # incomplete (missing score)

# Generated Performance Records 
records = []
for i, (s, sub, has_score) in enumerate(plan):
    score = make_score(i) if has_score else None
    records.append((students[s], subjects[sub], score))

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
print("No.  Student      Subject       Score")
for n, (st, sub, sc) in enumerate(records, 1):
    shown = sc if sc is not None else "MISSING"
    print(f"{n:<4} {st:<12} {sub:<13} {shown}")

# Organized Student/Subject Data (student/subject/scores)
organized = {}
for st, sub, sc in records:
    if st not in organized:
        organized[st] = {}
    if sub not in organized[st]:
        organized[st][sub] = []
    organized[st][sub].append(sc)

print("\n=== 2. ORGANIZED STUDENT/SUBJECT DATA ===")
for st in organized:
    print(st)
    for sub in organized[st]:
        print("   ", sub, "->", organized[st][sub])

# Repeated / Incomplete Records 
seen = set()
repeated = []
incomplete = []
for st, sub, sc in records:
    if (st, sub) in seen and (st, sub) not in repeated:
        repeated.append((st, sub))
    seen.add((st, sub))
    if sc is None:
        incomplete.append((st, sub))

print("\n=== 3. REPEATED / INCOMPLETE RECORDS ===")
print("Repeated student-subject records:", repeated)
print("Incomplete records (missing score):", incomplete)

# valid records only (records with a score)
valid = [r for r in records if r[2] is not None]

# Performance Categories 
print("\n=== 4. PERFORMANCE CATEGORIES ===")
categorized = []
for st, sub, sc in valid:
    cat = get_category(sc)
    categorized.append((st, sub, sc, cat))
    print(f"  {st:<12} {sub:<13} {sc:<4} {cat}")

category_count = {}
for st, sub, sc, cat in categorized:
    category_count[cat] = category_count.get(cat, 0) + 1
print("\nCategory count:")
for cat in ["EXCELLENT", "VERY GOOD", "GOOD", "FAIR", "POOR"]:
    print(" ", cat, "=", category_count.get(cat, 0))

# Student and Subject Summaries 
student_scores = {}
subject_scores = {}
for st, sub, sc in valid:
    student_scores.setdefault(st, []).append(sc)
    subject_scores.setdefault(sub, []).append(sc)

print("\n=== 5. STUDENT AND SUBJECT SUMMARIES ===")
print("Per student:")
student_avg = {}
for st in students:
    vals = student_scores[st]
    student_avg[st] = round(sum(vals) / len(vals), 2)
    print(f"  {st:<12} records={len(vals)} total={sum(vals)} "
          f"highest={max(vals)} lowest={min(vals)} average={student_avg[st]}")

print("\nPer subject:")
subject_avg = {}
for sub in subjects:
    vals = subject_scores[sub]
    subject_avg[sub] = round(sum(vals) / len(vals), 2)
    print(f"  {sub:<13} records={len(vals)} total={sum(vals)} "
          f"highest={max(vals)} lowest={min(vals)} average={subject_avg[sub]}")

# ---------- 6. Ordered Results ----------
print("\n=== 6. ORDERED RESULTS ===")
print("Valid records, highest to lowest score:")
for st, sub, sc, cat in sorted(categorized, key=lambda r: r[2], reverse=True):
    print(f"  {sc:<4} {st:<12} {sub:<13} {cat}")

print("\nStudents ranked by average:")
rank = 1
for st in sorted(student_avg, key=student_avg.get, reverse=True):
    print(f"  {rank}. {st:<12} {student_avg[st]}")
    rank += 1

print("\nSubjects ranked by average:")
rank = 1
for sub in sorted(subject_avg, key=subject_avg.get, reverse=True):
    print(f"  {rank}. {sub:<13} {subject_avg[sub]}")
    rank += 1

# ---------- FINAL STUDENT PERFORMANCE REPORT ----------
all_scores = [sc for st, sub, sc in valid]
best = max(valid, key=lambda r: r[2])
worst = min(valid, key=lambda r: r[2])
top_student = max(student_avg, key=student_avg.get)
top_subject = max(subject_avg, key=subject_avg.get)

print("\n" + "=" * 62)
print("FINAL STUDENT PERFORMANCE REPORT")
print("=" * 62)
print("Student :", LAST_NAME, "|", SEED_NUM, "|", FAVORITE_ARTIST)
print("Total records    :", len(records))
print("Valid records    :", len(valid))
print("Repeated record  :", repeated)
print("Incomplete record:", incomplete)
print("Overall average  :", round(sum(all_scores) / len(all_scores), 2))
print("Highest score    :", best)
print("Lowest score     :", worst)
print("Top student      :", top_student, "(", student_avg[top_student], ")")
print("Top subject      :", top_subject, "(", subject_avg[top_subject], ")")
print("-" * 62)
print(f"{'STUDENT':<12} {'SUBJECT':<13} {'SCORE':<8} CATEGORY")
for st, sub, sc in records:
    if sc is None:
        print(f"{st:<12} {sub:<13} {'MISSING':<8} INCOMPLETE")
    else:
        print(f"{st:<12} {sub:<13} {sc:<8} {get_category(sc)}")
print("=" * 62)
```

```python
# Lab 4 Appendix - Additional Exercise 3: Transaction History Analyzer
# Student-specific inputs
LAST_NAME = "GERMAN"
SEED_NUM = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

seed_digits = [int(ch) for ch in SEED_NUM if ch.isdigit()]          # [3,6,4,3,3,6]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]     # U,N,I,Q,...

customers = [LAST_NAME + "_C" + str(n) for n in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

# Fixed prices (in pesos)
prices = {"NOTEBOOK": 45, "PEN": 15, "USB": 350, "HEADPHONES": 800, "CALCULATOR": 450}


def make_quantity(i):
    digit = seed_digits[i % len(seed_digits)]
    letter = artist_letters[i % len(artist_letters)]
    return 1 + (digit * (i + 1) + (ord(letter) - 64)) % 10

# (customer index, product index) for each of the 12 transactions
plan = [(0, 0), (1, 1), (2, 2), (3, 3), (4, 4),   # 5 base transactions
        (0, 0),                                    # repeated customer-product
        (0, 2), (1, 3), (2, 4), (3, 0), (4, 1),
        (1, 2)]

# Generated Transaction Records 
records = []
for i, (c, p) in enumerate(plan):
    records.append((customers[c], products[p], make_quantity(i)))

print("=== 1. GENERATED TRANSACTION RECORDS ===")
print("No.  Customer     Product       Qty")
for n, (c, p, q) in enumerate(records, 1):
    print(f"{n:<4} {c:<12} {p:<13} {q}")

# Organized Transaction Data (customer/product/quantities) 
organized = {}
for c, p, q in records:
    if c not in organized:
        organized[c] = {}
    if p not in organized[c]:
        organized[c][p] = []
    organized[c][p].append(q)

print("\n=== 2. ORGANIZED TRANSACTION DATA ===")
for c in organized:
    print(c)
    for p in organized[c]:
        print("   ", p, "(P" + str(prices[p]) + ") -> qty", organized[c][p])

# Repeated / Distinct Transactions
seen = set()
repeated = []
for c, p, q in records:
    if (c, p) in seen and (c, p) not in repeated:
        repeated.append((c, p))
    seen.add((c, p))

distinct_customers = set(c for c, p, q in records)
distinct_products = set(p for c, p, q in records)

print("\n=== 3. REPEATED / DISTINCT TRANSACTIONS ===")
print("Repeated customer-product transactions:", repeated)
print("Distinct customers:", len(distinct_customers), sorted(distinct_customers))
print("Distinct products :", len(distinct_products), sorted(distinct_products))
print("Distinct customer-product pairs:", len(seen), "of", len(records), "records")

# Transaction Values (quantity/price)
transactions = []
for c, p, q in records:
    transactions.append((c, p, q, prices[p], q * prices[p]))

print("\n=== 4. TRANSACTION VALUES ===")
print("No.  Customer     Product       Qty  Price  Value")
for n, (c, p, q, pr, v) in enumerate(transactions, 1):
    print(f"{n:<4} {c:<12} {p:<13} {q:<4} {pr:<6} {v}")
grand_total = sum(t[4] for t in transactions)
print("Overall dataset value: P" + str(grand_total))

# Customer Summaries 
customer_value = {}
customer_qty = {}
customer_count = {}
for c, p, q, pr, v in transactions:
    customer_value[c] = customer_value.get(c, 0) + v
    customer_qty[c] = customer_qty.get(c, 0) + q
    customer_count[c] = customer_count.get(c, 0) + 1

print("\n=== 5. CUSTOMER SUMMARIES ===")
for c in customers:
    print(f"  {c:<12} transactions={customer_count[c]} items={customer_qty[c]} "
          f"total=P{customer_value[c]}")

# Product Summaries 
product_value = {}
product_qty = {}
product_count = {}
for c, p, q, pr, v in transactions:
    product_value[p] = product_value.get(p, 0) + v
    product_qty[p] = product_qty.get(p, 0) + q
    product_count[p] = product_count.get(p, 0) + 1

print("\n=== 6. PRODUCT SUMMARIES ===")
for p in products:
    print(f"  {p:<12} price=P{prices[p]:<4} transactions={product_count[p]} "
          f"units sold={product_qty[p]} total=P{product_value[p]}")

# Ordered Results
print("\n=== 7. ORDERED RESULTS ===")
print("Transactions, highest to lowest value:")
for c, p, q, pr, v in sorted(transactions, key=lambda t: t[4], reverse=True):
    print(f"  P{v:<6} {c:<12} {p:<13} qty {q}")

print("\nCustomers ranked by total value:")
rank = 1
for c in sorted(customer_value, key=customer_value.get, reverse=True):
    print(f"  {rank}. {c:<12} P{customer_value[c]}")
    rank += 1

print("\nProducts ranked by total value:")
rank = 1
for p in sorted(product_value, key=product_value.get, reverse=True):
    print(f"  {rank}. {p:<12} P{product_value[p]}")
    rank += 1

# Overall Transaction Summary 
total_qty = sum(t[2] for t in transactions)
top_customer = max(customer_value, key=customer_value.get)
top_product = max(product_value, key=product_value.get)
biggest = max(transactions, key=lambda t: t[4])
smallest = min(transactions, key=lambda t: t[4])

print("\n=== 8. OVERALL TRANSACTION SUMMARY ===")
print("Total transactions :", len(transactions))
print("Total items sold   :", total_qty)
print("Total sales value  : P" + str(grand_total))
print("Average per transaction: P" + str(round(grand_total / len(transactions), 2)))
print("Top customer :", top_customer, "(P" + str(customer_value[top_customer]) + ")")
print("Top product  :", top_product, "(P" + str(product_value[top_product]) + ")")
print("Largest transaction :", biggest[0], biggest[1], "P" + str(biggest[4]))
print("Smallest transaction:", smallest[0], smallest[1], "P" + str(smallest[4]))

# ---------- FINAL CONSOLIDATED TRANSACTION REPORT ----------
print("\n" + "=" * 66)
print("FINAL CONSOLIDATED TRANSACTION REPORT")
print("=" * 66)
print("Student :", LAST_NAME, "|", SEED_NUM, "|", FAVORITE_ARTIST)
print("Transactions :", len(transactions), "| Items sold:", total_qty,
      "| Total sales: P" + str(grand_total))
print("Repeated purchase :", repeated)
print("Top customer :", top_customer, "| Top product:", top_product)
print("-" * 66)
print(f"{'CUSTOMER':<12} {'PRODUCT':<13} {'QTY':<4} {'PRICE':<6} {'VALUE'}")
for c, p, q, pr, v in sorted(transactions, key=lambda t: (t[0], t[1])):
    print(f"{c:<12} {p:<13} {q:<4} {pr:<6} {v}")
print("-" * 66)
print(f"{'CUSTOMER TOTALS':<20}")
for c in customers:
    print(f"  {c:<12} P{customer_value[c]}")
print(f"{'PRODUCT TOTALS':<20}")
for p in products:
    print(f"  {p:<12} P{product_value[p]}")
print("=" * 66)
```
